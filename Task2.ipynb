```python
#load important libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#load the dataset
df = pd.read_csv(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task2\Iris.csv");
```

```python
#first five rows
df.head()
```

```python
#check the datatype of the all columns
df.info()
```

```python
#create bar chart for the visulization
avg_data = df.groupby("Species").mean(numeric_only=True)

x = np.arange(len(avg_data.index))

width = 0.2

plt.figure(figsize=(10,6))

plt.bar(
    x - 1.5*width,
    avg_data['SepalLengthCm'],
    width,
    label='sepal length (cm)'
)

plt.bar(
    x - 0.5*width,
    avg_data['SepalWidthCm'],
    width,
    label='sepal width (cm)'
)

plt.bar(
    x + 0.5*width,
    avg_data['PetalLengthCm'],
    width,
    label='petal length (cm)'
)

plt.bar(
    x + 1.5*width,
    avg_data['PetalWidthCm'],
    width,
    label='petal width (cm)'
)

plt.title("Average Measurements by Species")
plt.xlabel("species")
plt.ylabel("Measurement (cm)")

plt.xticks(x, avg_data.index, rotation=45)

plt.legend()

plt.savefig(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task2\bar_chart.png");

plt.show()

```

```python
# Histogram of Sepal Length
plt.figure(figsize=(6,5))

plt.hist(df['SepalLengthCm'], bins=10)

plt.title("Histogram of Sepal Length")

plt.xlabel("Sepal Length (cm)")

plt.ylabel("Frequency")

plt.savefig(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task2\histogram.png")

plt.show()
```

```python
#scatter plot
plt.figure(figsize=(8,5))

plt.scatter(
    df['SepalLengthCm'],
    df['PetalLengthCm']
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")

plt.savefig(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task2\scatter_plot.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

species = df['Species'].unique()

for sp in species:
    
    data = df[df['Species'] == sp]
    
    plt.scatter(
        data['PetalLengthCm'],
        data['PetalWidthCm'],
        label=sp
    )

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.legend()

plt.savefig(r"C:\Users\Admin\OneDrive\Documents\Internship project\Task2\feature_comparison.png")

plt.show()
```
